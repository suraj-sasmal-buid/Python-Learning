#### String Slicing

**Syntax - string[start:sop:step]**

```python
name = "Suraj Sasmal"
print(name[0])
print(name[2])
print(name[3])
print(name[6])
```

```python
### Positive Indexing

name = "Python Programming"

print(name[0])
print(name[1])
print(name[2])
print(name[3])
print(name[4])
print(name[5])
print(name[6])
print(name[7])
print(name[8])
print(name[9])
print(name[10])
print(name[11])
print(name[12])
print(name[13])
print(name[14])
print(name[15])
print(name[16])
print(name[17])
```

```python
## Negative Indexing -> string[-1:]

name = "Python Programming"

print(name[-1])
print(name[-2])
print(name[-3])
print(name[-4])
print(name[-5])
print(name[-6])
print(name[-7])
print(name[-8])
print(name[-9])
print(name[-10])
print(name[-11])
print(name[-12])
print(name[-13])
print(name[-14])
print(name[-15])
print(name[-16])
print(name[-17])
print(name[-18])
```

```python
name = "Suraj Sasmal"

print(name[-1])
print(name[-2])
print(name[-3])
print(name[-4])
print(name[-5])
print(name[-6])
print(name[-7])
print(name[-8])
print(name[-9])
print(name[-10])
print(name[-11])
print(name[-12])
```

```python
## String[:stop]

name = "Suraj Sasmal"

print(name[:7])
```

```python
text = "Hello Welcome to the world of Python Programming"
print(text[:30])
```

```python
name = "Hinata Shoyo"
print(name[:5])
```

```python
## String[start:]

name = "Suraj Sasmal"
print(name[3:])
```

```python
name = "Uzumaki Naruto"
print(name[4:])
```

```python
## Reverse a string -> (string[::-1])

name = "Suraj Sasmal"
print(name[::-1])
```

```python
name = "Shoyo Hinata"
print(name[::-1])
```

```python
## Step Slicing 

name = "Suraj Sasmal"

print(name[1:11:2])
```

```python
text = "Welcome to the world of python programming"

print(text[1:30:2])
```
