#### Decleration of Strings

```python
# single quote decleration 

name = 'Tanjiro Kamado'
print(name)
```

```python
# double quote decleration 

name = "Hinata Shoyo"
print(name)
```

```python
## Triple single quotes 

name = '''Uzumaki Naruto'''
print(name)
```

```python
### Triple double quotes 
name = """Levi Ackerman"""
print(name)
```
