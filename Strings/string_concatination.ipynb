#### Concatenation 

```python
str1 = "Hello"
str2 = "World"

result = str1 + str2

print(result)
```

```python
str1 = "Hello"
str2 = "World"

result = str1 + " " + str2

print(result)
```

```python
first = "I"
second = "Love"
third = "Python"

result = first + " " + second + " " + third

print(result)
```

```python
## += 

name = "Suraj"

name += " Sasmal"

print(name)
```

```python
## Concatenation of string with variables 

name = "Suraj"

age = "20"

result = "My name is " + name + " and I am " + age + " years old."

print(result)
```
