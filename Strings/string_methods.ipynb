**enumerate()**

```python
fruits = ["Apple", "Banana", "Mango"]
result = enumerate(fruits)
print(list(result))
```

```python
fruits = ["Apple", "Banana", "Mango"]
for index, value in enumerate(fruits):
    print(index,value)
```

```python
fruits = ["Apple", "Banana", "Mango"]
for index, value in enumerate(fruits, start=1):
    print(index,value)
```
