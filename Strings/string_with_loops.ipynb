**Using String with for loop**

```python
name = "Suraj Sasmal"

for char in name:
    print(char)
```

```python
name = "Python"

for i in range(len(name)):
    print(i, name[i])
```

```python
text = "Hello, Welcome to the Python Programming"
count = 0

for char in text:
    if char in "aeiouAEIOU":
        count += 1
print("Total Vowels: ", count)
```

```python
name = "Suraj Sasmal"

for char in name:
    if char == "a":
        print("A is found")
    else:
        print(char)
```

**Using String in while loop**

```python
name = "Suraj Sasmal"
i = 0

while i <len(name):
    print(name[i])
    i += 1
```

```python
text = "Education"
i = 0

count = 0

while i < len(text):
    if text[i] in "aeiouAEIOU":
        count += 1
    i += 1
print("Total Vowels: ", count)
```

```python
text = "Python"
i = len(text) -1

while i >= 0:
    print(text[i], end="")
    i -= 1
```

**Using break, continue, and else with Strings**

```python
text = "Programming"

for char in text:
    if char == "a":
        break
    print(char)
```

```python
text = "Programming"

for char in text:
    if char == "a":
        continue
    print(char)
```

```python
text = "Python"

for char in text:
    print(char)
else:
    print("Loop Completed.")
```
